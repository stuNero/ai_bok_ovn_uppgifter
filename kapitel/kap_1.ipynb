# Faktafrågor

1. ***Hur hänger AI, ML och DL ihop?*** 
   > AI(artificiell intelligens) handlar om att man försöker få datorer att efterlikna intelligensen hos människor och djur. ML(maskininlärning) är en del utav AI där datorn kan lära sig utan att behöva bli programmerad för varenda grej den ska lära sig, den har istället en algoritm som låter den lära sig på egen hand. DL(Djupinlärning) är en del inom maskininlärning där själva inlärningsprocessen och processen för att producera utdata görs utav neurala nätverk vilket liknar mer hjärnan hos en människa eller ett djur.   

2. ***Vilka är de fyra problemkategorierna inom ML?*** 
	> 
	> 1. **Regression** 
	> vilket hanterar problem med kontinuerliga värden, värden på en skala såsom att gissa hur mycket ett hus är värt baserad på en viss plats med en viss storlek. 
	> 2. **Klassificering** 
	> vilket hanterar identifiering av olika diskreta värden. Detta kan vara t.ex. bildigenkänning där datorn ska gissa på om något är en hund, katt eller människa. 
	> 3. **Klustring** 
	> vilket hanterar grupperingen av datapunkter i ett dataset. Detta görs genom att man skapar en valfri mängd klustercentrar och de placeras i mitten av grupper med datapunkter vars längdenheter mellan varandra är som kortast. 
	> 4. **Dimensionsreducering**
	> vilket är ett verktyg som används för att optimera ett dataset genom att överföra dess dimensioner till ett lägre. Detta används för att göra träning för AI modeller snabbare.

	1. *Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?* 
	   > Träningsdatan behöver vara störst för att modellen ska kunna bli tränad på så många olika problem som möjligt, då blir den mest generisk. Valideringsdatan och testdatan kan vara mindre då man bara ska se vilken av modellerna som blivit tränad på datan som presterar eller predikterar bäst. Man tar dock senare all data, även validering och testdatan i träningen av den slutgiltigt valda modellen då man behöver all data man kan få för den färdiga produkten. Uppdelning brukar se ut som följande; 70-15-15 eller 60-20-20, etc. 
	2. *Vad är k-delad korsvalidering för något?*
	   > K-delad korsvalidering är ett sätt att träna och utvärdera en ai-modell på. Det utgår ifrån att man istället för att ta hela datasetet och dela upp det i träning, validering och testportioner en gång så gör man det under flera gånger där modellen tränas på ungefär 80% och sedan valideras på 20%. Under de olika iterationerna man tränar och validerar den så byts valideringsportionen utav datan ut till något som inte använts för validering tidigare. På detta sätt får man flera olika utvärderingar på modellerna istället för enbart en gång med det traditionella sättet. 
	3. *Vad är root mean squared error (RMSE) för något?* 
	   > Det är ett värde som påvisar hur felaktig en modells prediktioner är jämfört med de faktiska värdena i ett dataset. Det fungerar genom att ta alla predikterade värden och jämföra dom med alla faktiska värden. Sedan tar man detta upphöjt till 2 för att ta bort alla negativa värden då de inte mäter distans och kan motverka positiva värden i en summering. Efter detta tar man det genomsnittliga värdet för att få ett värde man kan jämföra med andra undersökningar och slutgiltligen tar man roten ur för att motverka upphöjningen för att få ett normalt värde igen. 
	4. *Vad är en hyperparameter för något? Vad är en parameter för något?* 
	   > En parameter är värden som modellen lär sig under träningsprocessen medans en hyperparameter bestämmer hur inlärningsprocessen ska bete sig. Ett exempel på hyperparameter är t.ex. delarna i k-delad korsvalidering, alltså att valideringsdatan ska vara just 20% eller 15%. 
	5. *Vad är grid search och hur hänger namnet “grid” och “search” ihop med processen som genomförs? Läser vi dokumentationen: https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV så ser vi att hyperparametern refit har standardvärdet True, vad innebär det?* 
	   > Grid search är ett automatiskt sätt att välja ut hyperparametrar på. Refit parametern bestämmer att modellen ska anpassa om sig med den bästa hyperparametern som den hittat och dataset som den fått.  
	6. *Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding och ordinal encoding.*
		> Skillnaden mellan `nominal` och `ordinal data` är om datan har en "naturlig" rangordning, det vill säga t.ex. att `nominal data` där svart, röd och gul inte kan rangordnas på nåt bra sätt medans `ordinal data` där 'missnöjd, neutral, nöjd' är rangordnat från minst nöjd till mest. Man kan som modell ingenjör dock införa en rangordning på `nominal data` genom att undersöka datan, t.ex. om färgen svart är dyrare än röd och röd är dyrare än gul. `One-hot-encoding` kan man använda för att skapa features eller kolumner utav denna `nominala datan`. `One-hot-encoding` tar en kolumn med uppkomst av färger och gör en kolumn för varje färg där raderna i kolumnen är 1 eller 0 för uppkomst eller ej. `Dummy-variable-encoding` gör samma sak som one-hot-encoding fast den droppar en kolumn som referens, detta för att den tredje kolumnen måste vara det tredje alternativet vid prediktioner då det inte är de 2 första kolumnerna. `Ordinal encoding` är att man anger 0 -> 1 -> $n$ till `ordinal data` vilket har en naturlig rangordning. 
	7. *Vad är feature engineering för något?*
	   >  Man väljer ut eller tar bort features som man som ingenjör anser är nödvändiga eller ej för modellens syfte. T.ex. antal kvm för ett hus kan man tänka sig påverka priset medans hur många ägare huset haft kanske inte alls har stor påverkan på priset. Man använder alltså sitt eget omdöme och viss teori(forskning på ämnet) för att göra dessa val. 
	8. *Vad menas med principle of parsimony?* 
	   > Det innebär att man inte ska ha en för komplex modell för ett problem som inte behöver det.

4. ***Vad menas med att “en modell är en förenkling av verkligheten”?*** 
   > Det innebär att en modell är bara tränad på vissa *features* eller egenskaper för den faktiska datan och kan därför omöjligen ha hela sanningen. 

5. ***Vad menas med att en modell är “överanpassad” eller overfitted på engelska?*** 
   > Den är anpassad för nära den riktiga datan. Detta innebär att den förlorar den generella trenden i datan och prediktionerna har därför hög felaktighet. 

6. ***Högre är bättre i scikit-learn scoring, vad innebär det?***
   > Scikit-learn standardiserar alla poängsättningar så att man enklare kan jämföra mellan olika typer av utvärderingar.

7. ***Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när respektive datakategori kan uppstå.***
   > - ***Tvärsnittsdata*** är som data av ett ögonblick i en observation utav flera individer. Det kan t.ex. vara data över befolkningen i flera städer.  
   > - ***Tidsseriedata*** är data som ordnats i tidsordning för att enklare förstå hur data för en specifik individ förändras över tid. Om man följer befolkningsexemplet kan det vara hur befolkningen i en stad förändras under 10 år. 
   > - ***Paneldata*** är en kombination av **Tvärsnittsdata** och **Tidsseriedata** där man följer en specifik individ över tid istället.
