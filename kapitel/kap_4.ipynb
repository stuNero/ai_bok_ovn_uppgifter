1. Vad kännetecknar klassificeringsproblem? Ge några exempel på tillämpningsområden.
> Klassificeringsproblem kännetecknas med ett problem att identifiera vad något är. Det vanligaste användningsområdet är nog bildigenkänning där modellen ska gissa på om något är en hund, katt eller människa t.ex. Ett annat är t.ex. prediktionen om en kund kommer att stanna kvar i företaget eller inte. Det finns lite olika typer utav klassificeringsmodeller;
    > - *Binär klassificering* som ovanstående exempel där man ska gissa på två utdata. 
    > - *Multiklass klassificering* som det första exemplet där modellen gissar på mer än två utdata. 
    > - *Multioutput klassificering* där modellen gissar på två eller fler binära områden. T.ex. om en person är lång eller kort och om den är gammal eller ung.  

2. Förklara hur OvR- och OvO-algoritmerna fungerar.
> - **OVR** eller one-vs-rest algoritmen kollar om en datapunkt är liknar de klasserna som den blivit tränad på. Detta görs genom att varje prediktion får ett sannolikhetspoäng där den med högst blir prediktionen modellen kör på. 
> - **OVO** eller One-vs-One där en modell tränas per klasspar där varje klass jämförs med alla andra. t.ex. i klasserna *a*, *b*, *c*, *d* så måste dessa jämföras: a=b, a=c, a=d, b=d, b=c,c=d. Vid en prediktion så körs alla modeller och den som får mest gissningar är den man väljer. 

3. Förklara följande utvärderingsmått:
    - a) Confusion matrix
        > Detta är en matris som visualiserar hur en modells prediktioner presterar, den kollar hur prediktionen för varje klass har skett, t.ex. hur många gånger den blivit predikterad som någon av de andra klasserna. Övre vänster till nedre höger diagonalen visar de korrekta prediktionerna. Med detta kan man se vilken klass där prediktionerna blir bäst och sämst. 
    - b) Accuracy
        > *Accuracy* är ett mått för hur många prediktioner som blev korrekta. Det är i procent mellan 0 till 100. 
    - c) Precision
        > *Precision* mäter hur många av modellens positiva prediktioner som är korrekta. 
    - d) Recall
        > *Recall* mäter hur många av de faktiskt positiva fallen som modellen lyckas hitta. Ett högt *recall* betyder alltså att modellen missar få positiva fall.
    - e) $F_1$-score
        > $F_1$-score är ett mått som kombinerar *precision* och *recall*. Det beräknas som det harmoniska medelvärdet mellan dessa två mått. Ett högt värde innebär att modellen både hittar många positiva fall och att många av de positiva prediktionerna är korrekta.
    - f) ROC-kurvan
        > ROC-kurvan visar förhållandet mellan *true positive rate* och *false positive rate* vid olika klassificeringströsklar. Den används för att se hur bra modellen kan skilja mellan olika klasser. Ju närmare det övre vänstra hörnet kurvan ligger, desto bättre är modellen. AUC används ofta för att sammanfatta ROC-kurvan, där ett högre värde är bättre.

4. Vad är precision-recall tradeoff för något?
> *Precision-recall tradeoff* innebär att man ofta måste välja mellan hög precision och högt recall. Om man höjer tröskeln för när modellen ska klassa något som positivt blir modellen mer säker på sina positiva prediktioner. Då kan precision öka, men samtidigt kan modellen missa fler positiva fall och recall minska. Om man istället sänker tröskeln ökar ofta recall, men precision kan då minska.

5. Några vanligt förekommande modeller för klassificeringsproblem är enligt nedan. Förklara översiktligt hur respektive modell fungerar. Läs även igenom respektive modells dokumentation, notera att du inte behöver förstå alla detaljer från dokumentationen men det är bra att ha läst igenom den.
    - a) [Logistisk regression](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html)
        > Logistisk regression försöker hitta ett samband mellan indata och sannolikheten för olika klasser. Den passar bra för linjära klassificeringsproblem och kan användas för att avgöra vilken klass en datapunkt tillhör.
    - b) [Support vector machines](https://scikit-learn.org/stable/modules/generated/sklearn.svm.SVC.html)
        > *Support vector machine* försöker hitta den bästa gränsen mellan olika klasser. Gränsen placeras så att avståndet till de närmaste datapunkterna blir så stort som möjligt. Med hjälp av kernels kan modellen även hantera mer komplicerade, icke-linjära samband.
    - c) [Beslutsträd](https://scikit-learn.org/stable/modules/generated/sklearn.tree.DecisionTreeClassifier.html)
        > Ett beslutsträd delar upp datan stegvis genom olika frågor om variablerna. Till slut hamnar datapunkten i ett löv som innehåller modellens prediktion. Träd är relativt lätta att tolka men kan lätt bli överanpassade.
    - d) Ensemble learning – [VotingClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.VotingClassifier.html) och [BaggingClassifier](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.BaggingClassifier.html)
        > *Ensemble learning* innebär att man kombinerar flera modeller för att få en bättre prediktion. *VotingClassifier* låter flera modeller rösta om vilken klass som ska väljas. *BaggingClassifier* tränar flera modeller på olika slumpmässiga urval av träningsdatan och kombinerar sedan deras resultat.
    - e) [Random forest](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestClassifier.html)
        > *Random forest* består av många beslutsträd. Varje träd tränas på ett slumpmässigt urval av datan och variablerna. Resultatet från träden kombineras sedan genom omröstning. Detta gör modellen mer stabil än ett enskilt beslutsträd.
    - f) [Extra trees](https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.ExtraTreesClassifier.html)
        > *Extra trees* liknar *random forest* eftersom modellen består av många beslutsträd. Skillnaden är att uppdelningarna i träden görs ännu mer slumpmässigt. Detta kan göra modellen snabb och minska risken för överanpassning.

6. Vad innebär det att vi kan kolla på feature importance med hjälp av trädmodeller såsom beslutsträd eller random forest?
    > *Feature importance* visar hur mycket de olika variablerna har bidragit till modellens beslut. I ett beslutsträd kan man undersöka vilka variabler som används mest för att dela upp datan. I en *random forest* kombineras betydelsen från alla träd. På så sätt kan man se vilka variabler som verkar vara viktigast för modellens prediktioner. Det betyder dock inte automatiskt att variablerna orsakar resultatet.

7. Stina säger till Kalle på lunchsamtalet: *"jag vill ha högsta möjliga precision för vår klassificieringsmodell"*. Kalle funderar ett tag och säger: *"men vad händer då med recall?"*. Vad hade du svarat? I vilka fall kan man tänka sig vilja ha en så hög precision som möjligt? I vilka fall kan det vara dåligt? Om vi tänker oss rättsväsendet där en slutgiltig dom kan leda till fängelse, vad kan vi då säga om precision-recall tradeoff?
    > Om precisionen ska vara så hög som möjligt så försöker modellen bara klassa något som positivt när den är väldigt säker. Detta leder oftast till att precisionen ökar, men recall minskar eftersom fler positiva fall missas. Hög precision kan vara bra när falska positiva svar får stora konsekvenser, t.ex. när ett spamfilter inte ska råka markera viktiga mejl som spam. Det kan däremot vara dåligt om det viktigaste är att hitta alla positiva fall, exempelvis vid sjukdomsdiagnoser där det är bättre att undersöka en person en gång extra än att missa en sjukdom. 

    > 
    > Inom rättsväsendet är det särskilt viktigt att ha hög precision eftersom en felaktig positiv prediktion kan innebära att en oskyldig person döms till fängelse. Samtidigt måste man tänka på att även ett lågt recall kan göra att skyldiga personer missas. Därför bör en klassificeringsmodell inte ensam användas för att fatta beslut om skuld, utan resultatet behöver granskas av människor och kompletteras med annan bevisning.

9. På sidan 209 står det: *"på träningsdatan använder vi `.fit_transform()`, på valideringsdatan och testdatan använder vi endast `.transform()`."* Förklara logiken bakom detta.

    > På träningsdatan använder man `.fit_transform()` eftersom preprocessorn först måste lära sig olika värden från träningsdatan, till exempel medelvärde och standardavvikelse vid skalning. Därefter transformeras träningsdatan med dessa värden. På valideringsdatan och testdatan använder man endast `.transform()` så att samma värden används där. Om man skulle använda `.fit_transform()` även på validerings- eller testdatan skulle preprocessorn lära sig från den datan. Då kan information läcka in i modellen och resultatet bli bättre än vad det egentligen borde vara.

Notera: det huvudsakliga arbetet blir alltså att preprocessa, det vill säga bearbeta bilderna som tas via mobilen så att modellen kan prediktera dem. I verkligheten är det ofta långt ifrån trivialt att genomföra preprocessing.

```python
# Imports
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from sklearn.ensemble import VotingClassifier
```

```python
def display_confusion_matrix(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    ConfusionMatrixDisplay(cm) .plot()
```

```python
# Load data and inspect it
mnist = fetch_openml('mnist_784', version=1, cache=True,as_frame=False)
X = mnist["data"] [:10000]
y = mnist["target"] [:10000].astype(np.uint8)
print (X.shape)
print (y.shape)
some_digit = X[0]
some_digit_image = some_digit.reshape(28, 28)
plt.imshow(some_digit_image, cmap=mpl.cm.binary)
print("True label for the plotted image is", y[0])
# print (mnist.DESCR)
```

```python
# Splitting data
X_train_val, X_test, y_train_val, y_test = train_test_split(X, y, test_size=2000, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=2000, random_state=42)
# Scaling data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)
import joblib

joblib.dump(scaler, "../models/kap_4_15_scaler.joblib")
```

```python
logreg_clf = LogisticRegression(max_iter=1000)
random_forest_clf = RandomForestClassifier(n_estimators=100,random_state=42)
extra_trees_clf = ExtraTreesClassifier(n_estimators=100, random_state=42)
named_estimators = [
("logreg_clf", logreg_clf),
("random_forest_clf", random_forest_clf),
("extra trees clf", extra_trees_clf)
]
voting_clf = VotingClassifier(named_estimators, voting='hard')
models = [logreg_clf, random_forest_clf, extra_trees_clf, voting_clf]
for model in models:
    model.fit(X_train, y_train)
print("Accuracy for each model")

model_names = ["Logistic Regression", "Random Forest", "Extra Trees", "Voting Classifier"]
for name, model in zip(model_names, models):
    score = model.score(X_val, y_val)
    print (f"{name:20s}: {score: .4f}")
```

```python
et_pred_val = extra_trees_clf.predict(X_val)
display_confusion_matrix(y_val, et_pred_val)
print(classification_report(y_val, et_pred_val))
```
