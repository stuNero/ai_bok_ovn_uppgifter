## 3. Vad är scikit-learn för något? Biblioteket följer några centrala designprinciper. Vilka är dessa? Vad är *estimators*, *predictors* och *transformers*?

>   Det är ett bibliotek för maskininlärningsalgoritmer och verktyg med öppen källkod(alltså att man kan se hur allt är kodat). 
>   - **Estimators** - Detta är ett objekt inom scikitlearn som lär sig, anpassar sig eller estimerar data genom .fit() metoden.  
>   - **Predictors** - Predictors bygger på estimators och innebär att ett estimator objekt som tränats på data kan prediktera eller gissa på den nya indatan.  
>   - **Transformers** - Transformers är estimators som har .transform() metoden. Detta görs för att all data inte kan användas av alla modeller och behöver därför ändras på för att modellen ska kunna lära sig utav den. 
>   ### Designprinciper:
>   - **Konsistens** - Alla tillvägagångsätt inom SKLearn följer samma mönster, från metoder i de olika klasserna till dokumentationen om dessa. 
>   - **Inspektion** - Man ska enkelt kunna inspektera en klass, dess metoder och modellers hyperparametrar. Man kan göra detta genom dokumentationen eller med `.get_params()` metoden.
>   - **Begränsning av klasser** - Klasser är reserverade för modellalgoritmer medans för data använder man Numpy arrays eller SciPy sparse matrices. Modellernas hyperparametrar som python strängar och siffror. 
>   - **Rimliga standardvärden för hyperparametrar** - SciKitlearn har rimliga standardvärden (default values) för hyperparametrar som gör att man inte behöver mata in värden om man inte behöver specifiera något. 

## 4. Vad är *TensorFlow* och *Keras*?

#### Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor

from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV
```

### **b)** Dela upp datasetet i X och y.

```python
y = df[["target"]]
X = df.drop(df.columns[-1],axis=1)
```

### **c)** Dela upp datan ytterligare, i ett träning-, ett validering-, och ett testset med ``train_test_split()``. Låt 20% av datan vara testdata och 15% av den återstående datan vara valideringsdata.

```python
x_train_full, x_test, y_train_full, y_test = train_test_split(X, y, test_size=0.2, random_state=14)
# x_train_full bör vara 160
print(x_train_full.info())
x_train, x_val, y_train, y_val = train_test_split(x_train_full, y_train_full, test_size=0.15, random_state=14)
# 15%(validering) av 160(train_full) är 24
print(y_val.info())
```

### **d)** Träna två valfria regressionsmodeller (exempelvis ``LinearRegression`` och ``DecisionTreeRegressor``) på träningsdatan. Notera, i kapitel 3 kommer vi lära oss mer om ``DecisionTreeRegressor``.

```python
linear = LinearRegression()
linear.fit(x_train, y_train)

tree = DecisionTreeRegressor(random_state=14)
tree.fit(x_train, y_train)
```

### **e)** Utvärdera modellerna på valideringsdatan.

```python
def RMSE(y_true, y_pred):
    # gör till en dimensionella vektorer
    y_true = np.array(y_true).flatten()
    y_pred = np.array(y_pred).flatten()
    error_sum = 0
    for i in range(len(y_true)):
        error_sum += (y_true[i] - y_pred[i]) ** 2
    mean_error = error_sum / len(y_true)
    rmse = np.sqrt(mean_error)
    return rmse
```

```python
## Utvärderar LinearRegression

linear_pred = linear.predict(x_val)
linear_rmse = RMSE(y_val, linear_pred)

# Utvärderar DecisionTree
hyper_params =  {'max_depth': (None, 10, 20, 30)}
grid = GridSearchCV(
    tree,
    hyper_params,
    scoring='neg_root_mean_squared_error',
    cv=5
)

grid.fit(x_train, y_train)
tree_pred = grid.predict(x_val)
tree_rmse = RMSE(y_val, tree_pred)
```

```python
print("LinearRegression RMSE: ", linear_rmse)
print("DecisionTree RMSE: ", tree_rmse)
print("")
print("best depth:", grid.best_params_)
pd.DataFrame(grid.cv_results_)
```

*Max depth 10 verkar vara bäst, medans alla andra hamnar på samma andraplats*
